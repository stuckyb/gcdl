## Current implementation



```python
def init_tileset(files, crs):
    """
        files: A sequence of paths to the tile files.
    """
    fpaths = []
    polys = []

    # Extract the spatial coverage of each tile.
    for fpath in files:
        print('hi')
        fh = rasterio.open(fpath)
        bbox = fh.bounds
        print("bbox:",bbox)

        fpaths.append(fpath)

        coords = [
            [bbox.left, bbox.top], [bbox.right, bbox.top],
            [bbox.right, bbox.bottom], [bbox.left, bbox.bottom],
            [bbox.left, bbox.top]
        ]
        polys.append(sg.Polygon(coords))

    polys = gpd.GeoSeries(polys, crs=crs)
    fpaths = pd.Series(fpaths)

    return polys, fpaths
```

```python
polys, fpaths = init_tileset(tile_paths, 'EPSG:4326')
```
